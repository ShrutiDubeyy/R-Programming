```r
install.packages("rstatix")
library(tidyverse)
library(rstatix)
library(effectsize)
```

```r
install.packages('palmerpenguins')
library(palmerpenguins)

data("penguins")
head(penguins)
tail(penguins)
str(penguins)
dim(penguins)
names(penguins)
```

```r
colSums(is.na(penguins))
sum(complete.cases(penguins))

penguins_clean <- penguins %>%
  drop_na()

head(penguins_clean)

dim(penguins_clean)

```

```r
mean(penguins_clean$body_mass_g)
mean(penguins_clean$flipper_length_mm)
mean(penguins_clean$bill_length_mm)
mean(penguins_clean$bill_depth_mm)
```

```r
median(penguins_clean$body_mass_g)
median(penguins_clean$flipper_length_mm)
median(penguins_clean$bill_length_mm)
median(penguins_clean$bill_depth_mm)
```

```r
sd(penguins_clean$body_mass_g)
sd(penguins_clean$flipper_length_mm)
sd(penguins_clean$bill_length_mm)
sd(penguins_clean$bill_depth_mm)
```

```r
var(penguins_clean$body_mass_g)
var(penguins_clean$flipper_length_mm)
var(penguins_clean$bill_length_mm)
var(penguins_clean$bill_depth_mm)
```

```r
min(penguins_clean$body_mass_g)
max(penguins_clean$body_mass_g)
```

```r
penguins_clean %>%
  group_by(species) %>%
  summarise(
    n = n(),
    mean_body_mass = mean(body_mass_g),
    median_body_mass = median(body_mass_g),
    sd_body_mass = sd(body_mass_g),
    min_body_mass = min(body_mass_g),
    max_body_mass = max(body_mass_g)
  )
```

```r
ggplot(penguins_clean, aes(x = body_mass_g)) + geom_histogram(bins = 30) + labs( title = "Distribution of Penguin Body Mass", x = "Body Mass (g)", y = "Frequency" )
```

```r
ggplot(penguins_clean, aes(x = species, y = body_mass_g)) + geom_boxplot() + labs( title = "Body Mass by Penguin Species", x = "Species", y = "Body Mass (g)" )
```

```r
ggplot(penguins_clean, aes(x = species, y = flipper_length_mm)) + geom_boxplot() + labs( title = "Flipper Length by Species", x = "Species", y = "Flipper Length (mm)" )
```

```r
ggplot(penguins_clean, aes(x = flipper_length_mm, y = body_mass_g, color = species)) + geom_point() + labs( title = "Relationship Between Flipper Length and Body Mass", x = "Flipper Length (mm)", y = "Body Mass (g)" )
```

```r
ggplot(penguins_clean, aes(x = flipper_length_mm, y = body_mass_g)) + geom_point() + geom_smooth(method = "lm") + labs( title = "Flipper Length vs Body Mass", x = "Flipper Length (mm)", y = "Body Mass (g)" )
```

```r
t.test(penguins_clean$body_mass_g,
       conf.level=0.95)
```

```r
penguins_clean %>% group_by(species) %>% summarise( n = n(), mean = mean(body_mass_g), sd = sd(body_mass_g), se = sd / sqrt(n), lower_ci = mean - qt(0.975, df = n - 1) * se, upper_ci = mean + qt(0.975, df = n - 1) * se )
```

```r
# H₀: Mean body mass of males = Mean body mass of females

# H₁: Mean body mass of males ≠ Mean body mass of females

t.test(
  body_mass_g ~ sex,
  data = penguins_clean
)
```

```r
anova_model <- aov( flipper_length_mm ~ species, data = penguins_clean )
```

```r
eta_squared(anova_model)
```

```r
shapiro.test(residuals(anova_model))
```

```r
plot(anova_model)
```

```r
levene_test( penguins_clean, flipper_length_mm ~ species )
```

```r
kruskal.test( flipper_length_mm ~ species, data = penguins_clean )
```

```r
pairwise.wilcox.test( penguins_clean$flipper_length_mm, penguins_clean$species, p.adjust.method = "bonferroni" )
```

```r
cor.test( penguins_clean$flipper_length_mm, penguins_clean$body_mass_g, method = "pearson" )
```

```r
cor.test( penguins_clean$flipper_length_mm, penguins_clean$body_mass_g, method = "spearman" )
```

```r
# Parametric test t.test( body_mass_g ~ sex, data = penguins_clean ) # Non-parametric test wilcox.test( body_mass_g ~ sex, data = penguins_clean )
```
